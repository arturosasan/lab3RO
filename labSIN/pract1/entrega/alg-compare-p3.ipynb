# Análisis de Algoritmos de Búsqueda
Comparación estadística entre A* MD + LC y Voraz Manhattan sobre n puzzles aleatorios.

Métricas a comparar:
- Coste de la solución (número de movimientos)

- Tiempo de ejecución

```python
# Importar librerías necesarias
import search
from search import *
import time
import statistics
import matplotlib.pyplot as plt
```

## Configuración y Ejecución del Experimento

> **Ejemplo original del boletín (no es la Actividad 7).**

```python
# Configuración
size = 3  # Tamaño del puzzle
num_iterations = 1000  # Número de puzzles aleatorios
end_state = generate_spiral_state(size)

# Estructuras para almacenar resultados
astar_costs = []
astar_expanded = []
astar_times = []
greedy_costs = []
greedy_expanded = []
greedy_times = []
cost_differences = [] 
time_differences = [] 

print(f"Iniciando experimento con {num_iterations} puzzles {size}x{size}")
print("Estado objetivo:")
visualize_state(end_state)

# Bucle principal del experimento
for i in range(num_iterations):
    # Generar estado inicial aleatorio
    initial_state = generate_random_state(size, end_state)
    
    if not isSolvable(initial_state, end_state):
        print(f"\nPuzzle {i+1}: No tiene solución, generando otro...")
        continue
        
    print(f"\rProcesando puzzle {i+1}/{num_iterations}", end='', flush=True)
    
    # A* MD + LC

    result = run_algorithm('A (MD + LC)', initial_state, end_state, size)

    astar_result = result['A (MD + LC)']
    
    # Voraz Manhattan

    result = run_algorithm('Voraz (Manhattan)', initial_state, end_state, size)

    greedy_result = result['Voraz (Manhattan)']
    
    # Guardar resultados

    astar_costs.append(astar_result['cost'])
    astar_expanded.append(astar_result['nodes_expanded'])
    astar_times.append(astar_result['time'])
    
    greedy_costs.append(greedy_result['cost'])
    greedy_expanded.append(greedy_result['nodes_expanded'])
    greedy_times.append(greedy_result['time'])
    
    cost_diff = greedy_result['cost'] - astar_result['cost']
    cost_differences.append(cost_diff)
        
    time_diff = astar_result['time'] - greedy_result['time']
    time_differences.append(time_diff)
        


print("\nExperimento completado!")
```

## Análisis de Resultados

```python
# Calcular estadísticas
stats = {
    'A*': {
        'coste_min': min(astar_costs),
        'coste_max': max(astar_costs),
        'coste_medio': statistics.mean(astar_costs),
        'expandidos_medio': statistics.mean(astar_expanded),
        'tiempo_medio': statistics.mean(astar_times)
    },
    'Voraz': {
        'coste_min': min(greedy_costs),
        'coste_max': max(greedy_costs),
        'coste_medio': statistics.mean(greedy_costs),
        'expandidos_medio': statistics.mean(greedy_expanded),
        'tiempo_medio': statistics.mean(greedy_times)
    }
}

print("Estadísticas de A* MD LC:")
print(f"Coste - Min: {stats['A*']['coste_min']}, Max: {stats['A*']['coste_max']}, Media: {stats['A*']['coste_medio']:.2f}")
print(f"Nodos expandidos (media): {stats['A*']['expandidos_medio']:.2f}")
print(f"Tiempo medio: {stats['A*']['tiempo_medio']:.3f}s\n")

print("Estadísticas de Voraz Manhattan:")
print(f"Coste - Min: {stats['Voraz']['coste_min']}, Max: {stats['Voraz']['coste_max']}, Media: {stats['Voraz']['coste_medio']:.2f}")
print(f"Nodos expandidos (media): {stats['Voraz']['expandidos_medio']:.2f}")
print(f"Tiempo medio: {stats['Voraz']['tiempo_medio']:.3f}s\n")

print("Diferencias de coste (Voraz - A*):")
print(f"Mínima: {min(cost_differences)}")
print(f"Máxima: {max(cost_differences)}")
print(f"Media: {statistics.mean(cost_differences):.2f}")
```

## Visualización de Resultados

```python
# Gráfico de cajas para costes
plt.figure(figsize=(10, 6))
plt.boxplot([astar_costs, greedy_costs])
plt.title('Distribución de Costes por Algoritmo')
plt.ylabel('Número de movimientos')
plt.xticks([1, 2], ['A*', 'Voraz'])  # ← Aquí estableces las etiquetas
plt.show()


# Gráfico de barras para promedios
metrics = ['Coste Medio', 'Nodos Expandidos\n(media)', 'Tiempo Medio (s)']
astar_values = [stats['A*']['coste_medio'], stats['A*']['expandidos_medio'], stats['A*']['tiempo_medio']]
greedy_values = [stats['Voraz']['coste_medio'], stats['Voraz']['expandidos_medio'], stats['Voraz']['tiempo_medio']]

plt.figure(figsize=(12, 6))
x = range(len(metrics))
width = 0.35

plt.bar([i - width/2 for i in x], astar_values, width, label='A*', color='skyblue')
plt.bar([i + width/2 for i in x], greedy_values, width, label='Voraz', color='lightcoral')

plt.xlabel('Métricas')
plt.ylabel('Valores')
plt.title('Comparación de Métricas entre Algoritmos')
plt.xticks(x, metrics)
plt.legend()
plt.show()

# Histograma de diferencias de coste
plt.figure(figsize=(10, 6))
plt.hist(cost_differences, bins=20, edgecolor='black')
plt.title('Distribución de Diferencias de Coste (Voraz - A*)')
plt.xlabel('Diferencia en número de movimientos')
plt.ylabel('Frecuencia')
plt.show()

# Histograma de diferencias de tiempo
plt.figure(figsize=(10, 6))
plt.hist(time_differences, bins=100, edgecolor='black')
plt.title('Distribución de Diferencias de Tiempo (A* - Voraz)')
plt.xlabel('Diferencia en unidades de tiempo')
plt.ylabel('Frecuencia')
plt.show()
```

## Actividad 7 — Comparación de las dos mejores heurísticas

Según la Actividad 6 de `puzzle-p2.ipynb`, las dos mejores son:

- **`A (MD + LC)`**: la que menos nodos expande y menos memoria usa entre las heurísticas admisibles, manteniendo el coste óptimo.
- **`A* (Manhattan)`**: la más rápida entre las heurísticas admisibles, también con coste óptimo.

`Nueva Heurística` no se considera porque no es admisible. En este experimento se comparan las dos heurísticas sobre 1 000 puzzles 3×3 aleatorios, midiendo la **calidad de la solución (coste)**, los **nodos expandidos**, la **memoria** (`max_nodes`) y el **tiempo de ejecución**.

```python
# Configuración del experimento
size = 3
num_iterations = 1000
end_state = generate_spiral_state(size)

# Estructuras para almacenar resultados
manhattan_costs = []
manhattan_expanded = []
manhattan_max_nodes = []
manhattan_times = []
mdlc_costs = []
mdlc_expanded = []
mdlc_max_nodes = []
mdlc_times = []

for i in range(num_iterations):
    # generate_random_state ya garantiza que el estado sea resoluble
    initial_state = generate_random_state(size, end_state)

    # A* con la heurística de Manhattan
    result = graphSearch(initial_state, end_state, lambda _: 1,
                         getManhattanDistance, size)
    manhattan_costs.append(result['cost'])
    manhattan_expanded.append(result['nodes_expanded'])
    manhattan_max_nodes.append(result['max_nodes'])
    manhattan_times.append(result['time'])

    # A* con la heurística MD + LC
    result = graphSearch(initial_state, end_state, lambda _: 1,
                         get_md_plus_linear_conflict, size)
    mdlc_costs.append(result['cost'])
    mdlc_expanded.append(result['nodes_expanded'])
    mdlc_max_nodes.append(result['max_nodes'])
    mdlc_times.append(result['time'])

# Diferencias entre las dos heurísticas
cost_differences = [m - l for m, l in zip(manhattan_costs, mdlc_costs)]
time_differences = [l - m for l, m in zip(mdlc_times, manhattan_times)]
```

### Estadísticas

```python
# Calcular estadísticas de las dos heurísticas
stats = {
    'A* (Manhattan)': {
        'coste_min': min(manhattan_costs),
        'coste_max': max(manhattan_costs),
        'coste_medio': statistics.mean(manhattan_costs),
        'expandidos_medio': statistics.mean(manhattan_expanded),
        'memoria_media': statistics.mean(manhattan_max_nodes),
        'tiempo_medio': statistics.mean(manhattan_times),
    },
    'A (MD + LC)': {
        'coste_min': min(mdlc_costs),
        'coste_max': max(mdlc_costs),
        'coste_medio': statistics.mean(mdlc_costs),
        'expandidos_medio': statistics.mean(mdlc_expanded),
        'memoria_media': statistics.mean(mdlc_max_nodes),
        'tiempo_medio': statistics.mean(mdlc_times),
    },
}

for name, s in stats.items():
    print(f"Estadísticas de {name}:")
    print(f"Coste - Min: {s['coste_min']}, Max: {s['coste_max']}, "
          f"Media: {s['coste_medio']:.2f}")
    print(f"Nodos expandidos (media): {s['expandidos_medio']:.2f}")
    print(f"Memoria - max_nodes (media): {s['memoria_media']:.2f}")
    print(f"Tiempo medio: {s['tiempo_medio'] * 1000:.1f} ms\n")

print("Diferencias de coste (Manhattan - MD + LC):")
print(f"  Mínima: {min(cost_differences)}")
print(f"  Máxima: {max(cost_differences)}")
print(f"  Media: {statistics.mean(cost_differences):.2f}")

print("Diferencias de tiempo (MD + LC - Manhattan):")
print(f"  Mínima: {min(time_differences) * 1000:.3f} ms")
print(f"  Máxima: {max(time_differences) * 1000:.3f} ms")
print(f"  Media: {statistics.mean(time_differences) * 1000:.3f} ms")

print("¿Mismo coste en todos los puzzles?",
      manhattan_costs == mdlc_costs)
```

### Gráficos

```python
# Etiquetas de los algoritmos
etiquetas = ['A* (Manhattan)', 'A (MD + LC)']

# Distribución de costes
plt.figure(figsize=(8, 6))
plt.boxplot([manhattan_costs, mdlc_costs])
plt.title('Distribución de costes por algoritmo')
plt.ylabel('Número de movimientos')
plt.xticks([1, 2], etiquetas)
plt.show()

# Comparación de las métricas medias
metricas = [
    ('Coste medio', [stats['A* (Manhattan)']['coste_medio'],
                     stats['A (MD + LC)']['coste_medio']]),
    ('Nodos expandidos', [stats['A* (Manhattan)']['expandidos_medio'],
                          stats['A (MD + LC)']['expandidos_medio']]),
    ('Memoria (max_nodes)', [stats['A* (Manhattan)']['memoria_media'],
                             stats['A (MD + LC)']['memoria_media']]),
    ('Tiempo (ms)', [stats['A* (Manhattan)']['tiempo_medio'] * 1000,
                     stats['A (MD + LC)']['tiempo_medio'] * 1000]),
]
fig, axes = plt.subplots(1, 4, figsize=(18, 4))
for ax, (titulo, valores) in zip(axes, metricas):
    ax.bar(etiquetas, valores, color=['skyblue', 'lightcoral'])
    ax.set_title(titulo)
plt.tight_layout()
plt.show()

# Histograma de diferencias de coste
plt.figure(figsize=(10, 6))
plt.hist(cost_differences, bins=20, edgecolor='black')
plt.title('Distribución de diferencias de coste (Manhattan - MD + LC)')
plt.xlabel('Diferencia en número de movimientos')
plt.ylabel('Frecuencia')
plt.show()

# Histograma de diferencias de tiempo
plt.figure(figsize=(10, 6))
plt.hist([d * 1000 for d in time_differences], bins=50, edgecolor='black')
plt.title('Distribución de diferencias de tiempo (MD + LC - Manhattan)')
plt.xlabel('Diferencia en milisegundos')
plt.ylabel('Frecuencia')
plt.show()
```

## Declaración: comparación de las dos mejores heurísticas

Los datos anteriores corresponden a **1 000 puzzles 3×3** generados aleatoriamente y a una ejecución concreta; al no fijar semillas, los valores pueden variar ligeramente entre ejecuciones.

| Algoritmo | Calidad: coste medio (mín–máx) | Nodos expandidos (media) | Memoria: `max_nodes` (media) | Tiempo medio |
|---|---|---|---|---|
| **A\* (Manhattan)** | 21,54 (6–29) | 588,77 | 942,60 | **9,0 ms** |
| **A (MD + LC)** | 21,54 (6–29) | **307,86** | **491,52** | 19,9 ms |

- **Calidad de la solución (coste): empate.** Las dos obtienen el **mismo coste en los 1 000 puzzles** (diferencia mínima, máxima y media = 0) y el mismo coste medio **21,54** (mín. 6, máx. 29). Es el **óptimo**, porque ambas son admisibles y consistentes (Actividad 6).
- **Memoria y esfuerzo de búsqueda: gana `A (MD + LC)`.** Expande **307,86** nodos de media frente a **588,77** de `A* (Manhattan)` (**47,7 % menos**) y necesita **491,52** `max_nodes` frente a **942,60** (**47,9 % menos**).
- **Tiempo: gana `A* (Manhattan)`.** Tarda **9,0 ms** de media frente a **19,9 ms** de `A (MD + LC)`; es decir, `A (MD + LC)` necesita **≈2,2× más tiempo** (diferencia MD + LC − Manhattan: mín. −0,763 ms, máx. +83,827 ms, media **+10,846 ms**). Calcular los conflictos lineales cuesta más por nodo y ese sobrecoste supera el ahorro en expansiones.

En resumen, `A (MD + LC)` es mejor si el cuello de botella es la memoria o el número de expansiones, y `A* (Manhattan)` es mejor si prima el tiempo. **Las dos devuelven la solución óptima.**

## Conclusión

- **Calidad de la solución (coste):** empate; ambas devuelven el coste óptimo.
- **Memoria y esfuerzo de búsqueda:** gana `A (MD + LC)`.
- **Tiempo:** gana `A* (Manhattan)`.
